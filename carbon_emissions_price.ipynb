## Carbon Emissions Historical Price

```python
import pandas as pd
```

```python
df = pd.read_csv(
    "data/Carbon Emissions Futures Historical Data.csv",
    encoding="utf-8-sig",   
)
```

```python
df = df.drop(columns=["Open", "High", "Low", "Vol.", "Change %"])

df["Date"] = pd.to_datetime(df["Date"], format="%m/%d/%Y")
df = df.rename(columns={"Price": "EUA_PRICE_EUR"})

df["YEAR"]  = df["Date"].dt.year
df["MONTH"] = df["Date"].dt.month

monthly = (
    df.groupby(["YEAR", "MONTH"])["EUA_PRICE_EUR"]
    .mean()
    .round(2)
    .reset_index()
)

monthly["DATE"] = pd.to_datetime(
    monthly[["YEAR", "MONTH"]].assign(DAY=1)
).dt.strftime("%Y-%m-%d")

monthly = monthly[["DATE", "YEAR", "MONTH", "EUA_PRICE_EUR"]]
```

```python
monthly.to_csv("data/clean/carbon_price_monthly_cleaned.csv", index=False)
```
