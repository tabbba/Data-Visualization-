## 3. Missing Values Analysis

```python
def missing_report(df, name):
    miss = df.isnull().sum()
    pct  = (miss / len(df) * 100).round(2)
    report = pd.DataFrame({'missing_count': miss, 'missing_%': pct})
    report = report[report['missing_count'] > 0].sort_values('missing_%', ascending=False)
    print(f'\n=== {name} ===')
    print(report if not report.empty else 'No missing values.')

missing_report(df_workforce, 'global_ai_workforce')
missing_report(df_jobs,      'ai_job_dataset')
missing_report(df_trends,    'ai_job_trends')
```

## 6. Cleaning
### 6.1 Global AI Workforce

```python
wf = df_workforce.copy()

# make sure the Year is integer
wf['Year'] = wf['Year'].astype(int)

# Strip whitespace from Country
wf['Country'] = wf['Country'].str.strip()

# Confirm no out-of-range years
out_of_range = wf[~wf['Year'].between(2015, 2025)]
print(f'Rows outside 2015–2025: {len(out_of_range)}')
wf = wf[wf['Year'].between(2015, 2025)]

# Clip percentage columns to [0, 100]
pct_cols = ['Automation_Rate_Percent', 'Employment_Rate_Percent']
for c in pct_cols:
    if c in wf.columns:
        wf[c] = wf[c].clip(0, 100)

# feat engineering

# Net job change
wf['Net_Job_Change_Million'] = wf['Job_Creation_Million'] - wf['Job_Displacement_Million']

# Region mapping
region_map = {
    # Americas
    'United States': 'Americas',
    'Canada':        'Americas',
    'Brazil':        'Americas',
    'Mexico':        'Americas',
    # Europe
    'United Kingdom': 'Europe',
    'Germany':        'Europe',
    'France':         'Europe',
    'Italy':          'Europe',
    'Spain':          'Europe',
    'Netherlands':    'Europe',
    'Sweden':         'Europe',
    'Russia':         'Europe',
    # Asia
    'China':          'Asia',
    'Japan':          'Asia',
    'India':          'Asia',
    'South Korea':    'Asia',
    'Singapore':      'Asia',
    # Middle East & Africa
    'Turkey':         'Middle East & Africa',
    'South Africa':   'Middle East & Africa',
    # Oceania
    'Australia':      'Oceania',
}
wf['Region'] = wf['Country'].map(region_map).fillna('Other')

# Show any countries that fell into Other (update region_map above if needed)
unmapped = wf.loc[wf['Region'] == 'Other', 'Country'].unique()
if len(unmapped):
    print('Unmapped countries (add to region_map):', sorted(unmapped))

print(f'Cleaned workforce shape: {wf.shape}')
wf.head()
```

### 6.2 AI Job Dataset

```python
jobs = df_jobs.copy()



# Standardise text columns — strip and title-case
str_cols = ['job_title', 'industry', 'company_location', 'employee_residence',
            'experience_level', 'employment_type', 'company_size', 'education_required']
for c in str_cols:
    if c in jobs.columns:
        jobs[c] = jobs[c].str.strip().str.title()

# Salary: remove rows with non-positive salaries
before = len(jobs)
jobs = jobs[jobs['salary_usd'] > 0]
print(f'Removed {before - len(jobs)} rows with salary <= 0')

# Cap extreme salary outliers at 99th percentile
p99 = jobs['salary_usd'].quantile(0.99)
p01 = jobs['salary_usd'].quantile(0.01)
outliers_high = (jobs['salary_usd'] > p99).sum()
outliers_low  = (jobs['salary_usd'] < p01).sum()
print(f'Salary outliers above p99 ({p99:,.0f}): {outliers_high}')
print(f'Salary outliers below p01 ({p01:,.0f}): {outliers_low}')
jobs['salary_usd'] = jobs['salary_usd'].clip(p01, p99)

# remote_ratio — clip to [0, 100]
jobs['remote_ratio'] = jobs['remote_ratio'].clip(0, 100)



print(f'Cleaned jobs shape: {jobs.shape}')
jobs.head()
```

### 6.3 AI Job Trends

```python
trends = df_trends.copy()

# Standardise text columns
str_cols_t = ['Job Title', 'Industry', 'Job Status', 'AI Impact Level',
              'Required Education', 'Location']
for c in str_cols_t:
    if c in trends.columns:
        trends[c] = trends[c].str.strip().str.title()

# Rename columns for Power BI friendliness 
trends.columns = (
    trends.columns
    .str.strip()
    .str.replace(' ', '_')
    .str.replace(r'[^\w]', '', regex=True)
)
print('Renamed columns:', trends.columns.tolist())

# Clip percentage-like columns to [0, 100]
pct_like = [c for c in trends.columns if 'ratio' in c.lower() or 'pct' in c.lower()
            or 'percent' in c.lower() or 'risk' in c.lower() or 'diversity' in c.lower()]
for c in pct_like:
    trends[c] = trends[c].clip(0, 100)
    print(f'  clipped: {c}')

# Remove rows where Median_Salary is non-positive
salary_col = [c for c in trends.columns if 'salary' in c.lower()]
for c in salary_col:
    before = len(trends)
    trends = trends[trends[c] > 0]
    print(f'Removed {before - len(trends)} rows with {c} <= 0')

print(f'Cleaned trends shape: {trends.shape}')
trends.head()
```
